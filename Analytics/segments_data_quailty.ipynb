```python
import pandas as pd

total = len(df)
print(f"Всего строк в датасете: {total:,}")

# 1. Пропуски в profession и profession_resolved
p_na = df["profession"].isna().sum()
pr_na = df["profession_resolved"].isna().sum()
print(f"Пропусков в 'profession': {p_na:,} ({p_na/total*100:.2f}%)")
print(
    f"Пропусков в 'profession_resolved': {pr_na:,} ({pr_na/total*100:.2f}%)"
)

# 2. Как заполнялся profession_resolved?
print("\nОткуда брался profession_resolved (profession_source):")
print(df["profession_source"].value_counts(dropna=False))

# 3. Связь profession и segments
print("\nКросс-таблица: есть ли profession vs есть ли segments:")
print(
    pd.crosstab(
        df["profession"].isna(),
        df["segments"].isna(),
        rownames=["profession_is_NA"],
        colnames=["segments_is_NA"],
    )
)

print("\nТоп-10 segments, когда profession ПУСТАЯ:")
print(df[df["profession"].isna()]["segments"].value_counts(dropna=False).head(10))

# 4. Сравнение title и канонической profession
has_prof = df[df["profession"].notna()][["title", "profession"]].copy()

# Точное совпадение title == profession
exact = (
    has_prof["title"].str.strip().str.lower()
    == has_prof["profession"].str.strip().str.lower()
).mean() * 100
print(f"\nТочное совпадение 'title' == 'profession': {exact:.2f}%")

# Сколько уникальных title на одну каноническую профессию?
print("\nРазмытие: сколько уникальных title приходится на популярные профессии:")
top_profs = df["profession"].value_counts().head(5).index
for p in top_profs:
    n_titles = df[df["profession"] == p]["title"].nunique()
    total_p = (df["profession"] == p).sum()
    print(f"  • {p}: {total_p} вакансий и {n_titles} УНИКАЛЬНЫХ заголовков")

# 5. Примеры подмены профессии маркетингом в title
print("\nСлучайные примеры [title] -> [profession]:")
for _, r in has_prof.sample(10, random_state=42).iterrows():
    print(f"  Title: '{r['title']}'  --->  Profession: '{r['profession']}'")
```

```python
import re

print("=" * 60)
print("СТАТУС ОБОГАЩЕНИЯ")
print("=" * 60)
print("enrichment_status:")
print(df["enrichment_status"].value_counts(dropna=False))

print("\nenriched_fields (непустые значения):")
print(df["enriched_fields"].value_counts(dropna=False).head(5))

print("\nenrichment_version:")
print(df["enrichment_version"].value_counts(dropna=False).head(5))

# --------------------------------------------------------------------
# Замер плотности сигналов в description
# --------------------------------------------------------------------
# Чистим гомоглифы (латиницу в русских словах) перед regex
homoglyphs = {
    "a": "а",
    "c": "с",
    "e": "е",
    "o": "о",
    "p": "р",
    "x": "х",
    "y": "у",
    "k": "к",
    "m": "м",
    "t": "т",
    "b": "в",
    "h": "н",
}


def clean_text(t):
    if not isinstance(t, str):
        return ""
    t = t.lower()
    for lat, cyr in homoglyphs.items():
        t = t.replace(lat, cyr)
    return t


print("\nОбработка текстов описания...")
texts = df["description"].fillna("").apply(clean_text)

# Ключевые параметры
has_h12 = texts.str.contains(r"\b12\s*(?:час|ч\b)", regex=True).sum()
has_h8 = texts.str.contains(r"\b8\s*(?:час|ч\b)", regex=True).sum()
has_sched = texts.str.contains(r"\b(?:2/2|3/3|1/3|1/2|6/1|5/2)\b", regex=True).sum()
has_vahta = texts.str.contains(
    r"\b(?:15/15|30/15|45/15|60/30|30\s*дней|45\s*дней)\b", regex=True
).sum()

has_housing = texts.str.contains(
    r"\b(?:проживан|общежит|хостел|заселен)\w*", regex=True
).sum()
has_meals = texts.str.contains(r"\b(?:питан|обед|суточн)\w*", regex=True).sum()
has_uniform = texts.str.contains(r"\b(?:спецодежд|форм|сиз)\w*", regex=True).sum()
has_transfer = texts.str.contains(
    r"\b(?:развозк|транспорт|проезд|билет|гсм)\w*", regex=True
).sum()

# Отрицания
neg_housing = texts.str.contains(
    r"(?:без|не\s+предоставляется)\s+(?:проживан|общежит)", regex=True
).sum()
neg_duty = texts.str.contains(
    r"(?:не\s+нужно|без)\s+(?:разгруз|таскать|тяжест)", regex=True
).sum()

print("\n" + "=" * 60)
print(f"ПЛОТНОСТЬ СИГНАЛОВ В DESCRIPTION (всего {len(texts):,} текстов):")
print("=" * 60)
print(
    f"  • Смена 12 часов:              {has_h12:,} ({has_h12/total*100:.2f}%)"
)
print(
    f"  • Смена 8 часов:               {has_h8:,} ({has_h8/total*100:.2f}%)"
)
print(
    f"  • График (2/2, 3/3, 5/2, 6/1): {has_sched:,} ({has_sched/total*100:.2f}%)"
)
print(
    f"  • Цикл вахты (15/15, 30/15...):{has_vahta:,} ({has_vahta/total*100:.2f}%)"
)
print(
    f"  • Жилье / общежитие:           {has_housing:,} ({has_housing/total*100:.2f}%) [с отрицанием 'без жилья': {neg_housing}]"
)
print(
    f"  • Питание / суточные:          {has_meals:,} ({has_meals/total*100:.2f}%)"
)
print(
    f"  • Спецодежда / форма:          {has_uniform:,} ({has_uniform/total*100:.2f}%)"
)
print(
    f"  • Развозка / проезд / ГСМ:     {has_transfer:,} ({has_transfer/total*100:.2f}%)"
)
print(f"  • Отрицания обязанностей:      {neg_duty:,}")
```

```python
import pandas as pd

# 1. Анализ city_resolved
print("--- city_resolved_source ---")
if "city_resolved_source" in df.columns:
    print(df["city_resolved_source"].value_counts(dropna=False))

# 2. Срез: ВАХТА
# На Авито график вахты хранится в колонке 'График' или 'Режим работы' или выявляется по тексту
vahta_mask = (
    df["График"].astype(str).str.contains("вахт", case=False, na=False)
    | df["title"].str.contains("вахт", case=False, na=False)
    | (
        df["Длительность вахты"].notna()
        & (df["Длительность вахты"] != "не указано")
    )
)
vahta_df = df[vahta_mask]
print(
    f"\nВАХТА: {len(vahta_df)} вакансий ({len(vahta_df)/len(df)*100:.1f}% от всей базы)"
)
print("Топ-5 профессий на вахте (profession_resolved):")
print(vahta_df["profession_resolved"].value_counts().head(5))
print("\nТоп-5 городов размещения вахты (city_resolved):")
print(vahta_df["city_resolved"].value_counts().head(5))

# 3. Срез: УДАЛЁНКА
remote_mask = (
    df["Формат работы"].astype(str).str.contains("удален", case=False, na=False)
    | df["title"].str.contains("удален", case=False, na=False)
    | (df["city_resolved"].astype(str).str.contains("удален", case=False))
)
remote_df = df[remote_mask]
print(
    f"\nУДАЛЁНКА: {len(remote_df)} вакансий ({len(remote_df)/len(df)*100:.1f}% от всей базы)"
)
print("Топ-5 профессий на удалёнке (profession_resolved):")
print(remote_df["profession_resolved"].value_counts().head(5))

# 4. Проверка объема ключевых ячеек (profession_resolved x city_resolved)
valid_profs = [
    "Грузчик",
    "Продавец",
    "Кассир",
    "Водитель грузового транспорта",
    "Курьер",
    "Комплектовщик",
]
valid_cities = ["Москва", "Санкт-Петербург", "Екатеринбург"]

cell_matrix = (
    df[
        df["profession_resolved"].isin(valid_profs)
        & df["city_resolved"].isin(valid_cities)
    ]
    .groupby(["profession_resolved", "city_resolved"])
    .size()
    .unstack(fill_value=0)
)

print("\nМатрица наполненности ячеек [profession_resolved x city_resolved]:")
print(cell_matrix)
```

```python
import pandas as pd

# 1. ЧТО ЗА ПРОФЕССИИ НА УДАЛЁНКЕ И ВАХТЕ СПРЯТАНЫ В TITLE?
print("=" * 60)
print("1. КТО НА САМОМ ДЕЛЕ НА УДАЛЁНКЕ (Топ-10 заголовков):")
print("=" * 60)
print(remote_df["title"].value_counts().head(10))

print("\n" + "=" * 60)
print("2. КТО НА ВАХТЕ В UNRESOLVED (Топ-10 заголовков):")
print("=" * 60)
vahta_unresolved = vahta_df[vahta_df["profession_resolved"] == "unresolved"]
print(vahta_unresolved["title"].value_counts().head(10))

# 2. ПОЧЕМУ ПЕТРОПАВЛОВСК-КАМЧАТСКИЙ В ВАХТЕ?
print("\n" + "=" * 60)
print("3. ВАХТА В ПЕТРОПАВЛОВСКЕ-КАМЧАТСКОМ (Топ-5 заголовков и работодателей):")
print("=" * 60)
pk = vahta_df[vahta_df["city_resolved"] == "Петропавловск-Камчатский"]
print(pk[["title", "employer_name_observed"]].value_counts().head(5))

# 3. АНАЛИЗ SALARY_RAW ПО КЛЮЧЕВЫМ ЯЧЕЙКАМ (ПЕРИОДЫ ОПЛАТЫ)
print("\n" + "=" * 60)
print("4. ФОРМАТЫ ОПЛАТЫ В SALARY_RAW ПО НАШИМ 6 ПРОФЕССИЯМ:")
print("=" * 60)


def extract_salary_type(val):
    if pd.isna(val):
        return "missing"
    s = str(val).lower()
    if "смен" in s:
        return "за смену"
    if "час" in s:
        return "в час"
    if "день" in s or "дня" in s:
        return "в день"
    if "месяц" in s or "мес" in s:
        return "в месяц"
    return "без указания периода (число)"


df["salary_period_type"] = df["salary_raw"].apply(extract_salary_type)

# Смотрим долю "за смену" по нашим 6 профессиям
prof_salary_pivot = (
    pd.crosstab(
        df[df["profession_resolved"].isin(valid_profs)]["profession_resolved"],
        df["salary_period_type"],
        normalize="index",
    )
    * 100
).round(1)

print(prof_salary_pivot)

# 4. КАКИЕ ПАРАМЕТРЫ КАРТОЧКИ АВИТО РЕАЛЬНО ЗАПОЛНЕНЫ В НАШИХ 18 ЯЧЕЙКАХ?
print("\n" + "=" * 60)
print("5. ЗАПОЛНЕННОСТЬ СТРУКТУРИРОВАННЫХ ПОЛЕЙ В ЯЧЕЙКАХ (ТОП-6 ПРОФЕССИЙ):")
print("=" * 60)
core_cells_df = df[
    df["profession_resolved"].isin(valid_profs)
    & df["city_resolved"].isin(valid_cities)
]

attrs_to_check = [
    "График",
    "Занятость",
    "Опыт работы",
    "Частота выплат",
    "Сфера деятельности компании",
    "Способ оформления",
    "Количество рабочих часов в день",
    "Что получают работники",
    "Смены",
    "Медкнижка",
    "Категория водительских прав",
]

for attr in attrs_to_check:
    if attr in df.columns:
        pct = core_cells_df[attr].notna().mean() * 100
        print(f"  • {attr:32s}: {pct:5.1f}%")
```

## Выводы исследования признаков и пригодности сегментов (Зона Кости)

### 1. Разрешение профессий и непригодность заголовков (`profession_resolved` & `title`)
* **Иллюзия 100% заполненности:** Формально `profession_resolved` не имеет пропусков (0,0%), однако колонка `profession_source` показывает, что в **34,4% строк (7 917 шт.)** стоит статус `none` — каноническая роль не определена. Исходное поле `segments` не помогает восстановить пропуски (в 9 382 случаях оно `NaN`).
* **Тотальное размытие `title`:** Точное совпадение `title == profession` составляет всего **3,48%**. Заголовок является рекламным инструментом работодателя, а не названием должности:
  - На 2 830 грузчиков приходится 1 540 уникальных заголовков, на 2 409 водителей — 1 453 заголовка.
  - 21,4% заголовков содержат маркетинговый шум («Срочно», «Прямой работодатель», «выплаты по вторникам»), а также грубые смысловые сбои (например, объявление *«Курьер на электровело...»* каталогизировано платформой как *«Грузчик»*).
  - **Вывод для ML:** Использовать `title` как категориальную фичу категорически запрещено (вызывает взрыв размерности и утечку таргета). Для обучения пригодны только нормализованные профессии каталога (`profession_resolved != 'none'`).

### 2. Экспресс-оценка NLP-обогащения из `description`
* **Текущий статус обогащения:** В массиве из 23 036 строк поле `enriched_fields` не заполнено ни разу (100% `NaN`), а `enrichment_status: not_run` у 99,4% строк. Текстовое обогащение де-факто отсутствует.
* **Полезность текста:** Описание содержит критичные условия труда, которых нет в фильтрах карточки:
  - Формула графика (2/2, 3/3, 5/2, 6/1) явно прописана у **34,35% (7 912 строк)**.
  - Пакет компенсаций: питание упомянуто в **33,80% (7 787 строк)**, спецодежда — в **30,83% (7 102 строки)**, развозка/проезд — в **25,42% (5 856 строк)**, бесплатное жилье — в **19,68% (4 533 строки)**.
  - Длина смены (12 ч или 8 ч) явно зафиксирована у **9,72%**.
* **Трудоемкость и сложность:** В 94% описаний выявлены гомоглифы (латиница в русских словах для обхода антипарсинга), что требует обязательной де-гомоглифизации перед regex-парсингом. Проблема отрицаний («без проживания», «разгружать не нужно») присутствует в ~100–120 случаях, что надежно купируется контекстным окном $\pm 3$ токена. Дорогие LLM/NER на первом этапе избыточны: 85–90% полезных сигналов закрываются детерминированными регулярными выражениями.

### 3. Исследование изолированных срезов (Вахта и Удалёнка)
* **Вахта (20,4% базы, 4 694 вакансии):**
  - **Аномалия классификации:** 65,2% вахты (3 062 строки) не определены по профессии (`unresolved`), так как представлены неучтенными строительными и промышленными ролями (слесари, монтажники, охрана, разнорабочие).
  - **Аномалия гео:** В топ-4 городов вахты вошел Петропавловск-Камчатский (242 вакансии). Анализ работодателей («СВОих», «Работа контракт РФ», «ОТВАЖНЫЕ ЛЮДИ») показал, что это контрактная служба, пробившаяся через технический фильтр `military`.
  - **Вывод по вахте:** Вахта несовместима с первым бейзлайном из-за разрыва между местом найма и объектом работ, а также публикации дохода «за вахту 30–45 смен» (150–250 тыс. ₽). Сегмент исключается из первого этапа.
* **Удалёнка (2,7% базы, 624 вакансии):**
  - **Аномалия классификации:** 100% удаленки попало в `unresolved`, так как состоит из «белых воротничков» (менеджеры по продажам, рекрутеры, операторы колл-центров), которых нет в справочнике 6 целевых профессий.
  - **Вывод по удалёнке:** Сегмент не имеет географической привязки (`location_missing`) и исключается из первого обучения линейных моделей.
