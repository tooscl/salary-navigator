# Общий датасет → DataFrame

Запустите ячейки последовательно из корня репозитория или любого его подкаталога. Нужны Python, pandas и Jupyter. В Git хранятся части CSV и manifest; команда ниже проверяет их и восстанавливает локальный `Data/dataset/vacancies.csv`. **Она перезаписывает этот локальный файл**, поэтому сохраняйте результаты очистки и признаки отдельно.


```python
from pathlib import Path
import subprocess
import sys

# Ищем корень от рабочей папки ядра Jupyter, а не от расположения ноутбука.
cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents)
             if (p / "utils/build_dataset.py").is_file()
             and (p / "Data/dataset/manifest.json").is_file()), None)
if root is None:
    raise RuntimeError("Корень проекта T&L не найден. Запустите Jupyter из репозитория или его подкаталога.")

# То же, что python3 utils/build_dataset.py --materialize в терминале из корня.
subprocess.run([sys.executable, "utils/build_dataset.py", "--materialize"], cwd=root, check=True)
```

```python
import hashlib
import pandas as pd

csv_path = root / "Data/dataset/vacancies.csv"
with csv_path.open("rb") as f:
    digest = hashlib.sha256()
    for block in iter(lambda: f.read(1024 * 1024), b""):
        digest.update(block)
    input_sha256 = digest.hexdigest()
print(f"Вход: {csv_path}\nSHA-256: {input_sha256}")
df = pd.read_csv(csv_path, dtype={"id": "string"}, low_memory=False)
df.head()
```
